```python
import requests

def get_bibtex_from_doi(doi):
    """
    Retrieve BibTeX citation for a given DOI
    
    Args:
        doi (str): The DOI to look up
        
    Returns:
        str: BibTeX citation or error message
    """
    # Base URL for DOI resolution
    url = f'https://doi.org/{doi}'
    
    # Headers to request BibTeX format
    headers = {
        'Accept': 'application/x-bibtex',
        'User-Agent': 'Mozilla/5.0' # Some servers require a user agent
    }
    
    try:
        # Make the request
        response = requests.get(url, headers=headers)
        
        # Check if request was successful
        if response.status_code == 200:
            return response.text
        else:
            return f"Error: Unable to retrieve citation (Status code: {response.status_code})"
            
    except requests.exceptions.RequestException as e:
        return f"Error: {str(e)}"


# Example DOI
doi = "10.1038/s41586-021-03819-2"  # Replace with your DOI

bibtex = get_bibtex_from_doi(doi)
```
